```python

student=int(input("input student id:"))
score=int(input("input score: "))
# ตรวจสอบว่าคะแนนถูกต้องหรือไม่
def is_valid_score(score):
    #ตรวจสอบว่าคะแนนอยู่ระหว่าง 0 ถึง 100 หรือไม่
     return 0<= score <= 100
# แปลงคะแนนเป็นเกรด
def to_grade(score):
   #แปลงคะแนนเป็นเกรด A B C หรือ F
    if score >= 80:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    else:
        return "F"
# แสดงผลคะแนนนักศึกษา
def print_result(student_id, score):
    print("รหัสนักศึกษา :", student_id)
    print("คะแนน :", score)

    if is_valid_score(score):
        print("เกรด :", to_grade(score))
    else:
        print("คะแนนไม่ถูกต้อง")


print_result(student, score)



```

```python
# ฟังก์ชันคำนวณค่าเฉลี่ย
def safe_average(total, count):
    #คำนวณค่าเฉลี่ย ถ้า count ไม่เป็นบวกให้คืนค่า None
    if count <= 0:
        return None
    else:
        return round(total / count, 2)
# รับค่า PM2.5 จำนวน 5 วัน
day1 = float(input("วันที่ 1 : "))
day2 = float(input("วันที่ 2 : "))
day3 = float(input("วันที่ 3 : "))
day4 = float(input("วันที่ 4 : "))
day5 = float(input("วันที่ 5 : "))
# กำหนดตัวแปรสะสม
total = 0
count = 0
# ตรวจสอบและสะสมข้อมูล
if day1 >= 0:
    total = total + day1
    count = count + 1
if day2 >= 0:
    total = total + day2
    count = count + 1
if day3 >= 0:
    total = total + day3
    count = count + 1
if day4 >= 0:
    total = total + day4
    count = count + 1
if day5 >= 0:
    total = total + day5
    count = count + 1
#หาค่าเฉลี่ย
average = safe_average(total, count)

#แสดงผล
if average == None:
    print("ไม่มีข้อมูลที่ใช้ได้")
else:
    print("ผลรวม PM2.5 :", total)
    print("จำนวนวันที่มีข้อมูล :", count)
    print("ค่าเฉลี่ย PM2.5 :", average)
```
