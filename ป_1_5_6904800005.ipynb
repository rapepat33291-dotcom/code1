```python
#ป.1
scores = [78, 92, 65, 88, 45, 95, 72]

# นับจำนวนผู้เข้าสอบ
count = len(scores)

# หาคะแนนสูงสุดและต่ำสุด
high = max(scores)
low = min(scores)

# หาค่าเฉลี่ย
avg = sum(scores) / len(scores)

# นับผู้ที่ได้คะแนน 80 ขึ้นไป
passed80 = 0
for s in scores:
    if s >= 80:
        passed80 += 1

# เก็บคะแนนที่ไม่ผ่านเกณฑ์ 50
fail = []
for s in scores:
    if s < 50:
        fail.append(s)

# แสดงผล
print("จำนวนผู้เข้าสอบ:", count)
print("คะแนนสูงสุด:", high)
print("คะแนนต่ำสุด:", low)
print("คะแนนเฉลี่ย: {:.2f}".format(avg))
print("จำนวนผู้ได้ 80 ขึ้นไป:", passed80)
print("คะแนนที่ไม่ผ่านเกณฑ์:", fail)
```

```python
#ป.2
ids = ["C001", "C002", "C001", "C003", "C002", "C004"]

# สร้าง list ว่าง
unique = []

# ตรวจสอบรหัสที่ไม่ซ้ำ
for i in ids:
    if i not in unique:
        unique.append(i)

# คำนวณจำนวนรายการซ้ำ
total = len(ids)
not_duplicate = len(unique)
duplicate = total - not_duplicate

# แสดงผล
print("จำนวนรายการทั้งหมด:", total)
print("จำนวนรหัสที่ไม่ซ้ำ:", not_duplicate)
print("จำนวนรายการที่ซ้ำ:", duplicate)
print("รหัสที่ไม่ซ้ำ:", unique)
```

```python
#ป.3
payments = ["cash", "credit", "cash", "transfer", "cash", "credit"]

# สร้าง dict ว่าง
counter = {}

# นับจำนวนครั้ง
for p in payments:
    counter[p] = counter.get(p, 0) + 1

# เรียงตามชื่อวิธีชำระเงิน
for p in sorted(counter):
    print(p, ":", counter[p], "ครั้ง")

# หาวิธีที่ถูกใช้มากที่สุด
max_count = 0
max_payment = ""

for p in counter:
    if counter[p] > max_count:
        max_count = counter[p]
        max_payment = p

print("วิธีที่ใช้มากที่สุด:", max_payment)
print("จำนวนครั้ง:", max_count)
```

```python
#ป.4
orders = [
    {"id": 1001, "customer": "Alice", "city": "Bangkok", "product": "Notebook", "qty": 2, "price": 120},
    {"id": 1002, "customer": "Bob", "city": "Bangkok", "product": "Pen", "qty": 5, "price": 15},
    {"id": 1003, "customer": "Charlie", "city": "Chiang Mai", "product": "Backpack", "qty": 1, "price": 850},
    {"id": 1004, "customer": "Diana", "city": "Phuket", "product": "Notebook", "qty": 3, "price": 120},
    {"id": 1005, "customer": "Eve", "city": "Bangkok", "product": "Pen", "qty": 10, "price": 15},
    {"id": 1006, "customer": "Frank", "city": "Phuket", "product": "Backpack", "qty": 1, "price": 850}
]
# ยอดขายรวมทั้งหมด
total_sales = 0

# ยอดขายแยกตามสินค้า
product_sales = {}

# เก็บชื่อลูกค้า
customers = []

# เก็บยอดขายของลูกค้าในรูปแบบ tuple
pairs = []

for o in orders:

    # คำนวณยอดรวมของแต่ละคำสั่งซื้อ
    total = o["qty"] * o["price"]

    # รวมยอดขายทั้งหมด
    total_sales += total

    # รวมยอดขายแยกตามสินค้า
    product = o["product"]
    product_sales[product] = product_sales.get(product, 0) + total

    # เก็บชื่อลูกค้า
    customers.append(o["customer"])

    # เก็บยอดขายและชื่อลูกค้าเป็น tuple
    pairs.append((total, o["customer"]))

# เรียงลำดับยอดขายจากมากไปน้อย
pairs.sort(reverse=True)

# แสดงรายงาน
print("ยอดขายรวมทั้งหมด:", total_sales)

print("\nยอดขายแยกตามสินค้า:")
for p in sorted(product_sales):
    print(p, ":", product_sales[p])

print("\nจำนวนลูกค้าที่ไม่ซ้ำ:", len(set(customers)))

print("\nสามอันดับลูกค้าที่มียอดขายสูงสุด:")
for p in pairs[:3]:
    print(p[1], ":", p[0])
```
